# 7. NEURAL NETWORKS

## 7.1 Neural Networks

### 7.1.1 Neural Network Architecture

A classic neural network includes:

- Input Layer: receive Features;
- Hidden Layers: transform data with Weights, Biases and Activation Functions;
- Output Layer: produce the final output.

### 7.1.2 Forward Propagation

The calculation process of a neuron can be devided into 2 steps:

1. Linear Transformation: $z = w^T x +b$;
2. Activation Function: $a = f(z)$, in which f is the activation function.

Forward Propagation uses the output of a layer as the input of the next layer, repeat the process until the prediction comes out.

Let's take a look at an example:


```python
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


wi = np.array(
    [
        [0.25, 0.20],
        [0.30, 0.15],
    ]
)

wo = np.array(
    [
        [0.40, 0.45],
        [0.50, 0.55],
    ]
)

b1 = 0.30
b2 = 0.65
x = np.array([0.5, 0.1])

z1 = wi @ x + b1
a1 = sigmoid(z1)

z2 = wo @ a1 + b2
a2 = sigmoid(z2)

print("Hidden pre-activation:", z1)
print("Hidden activation:", a1)
print("Output pre-activation:", z2)
print("Final output:", a2)
```

notice: we agree on that:

$$
W.shape = (n_{\text{output neurons}},n_{\text{input neurons}})
$$


### 7.1.3 Activation Functions

Neural Networks need activation functions, because by linear algebra, the combination of several pure Linear layers is equal to a single Linear Layer. Thus, we need non-linear activation functions to learn more complex nonlinear relationships.

There are four important activation functions: Sigmoid, Tanh, ReLU and Softmax.


### 7.1.4 Loss Functions

Loss functions show the difference between model predictions and ground-truth tables.

- For Regression, the common output activation is Linear, and the common loss is MSE/MAE;
- For Binary Classification, the common output activation is Sigmoid, and the common loss is Binary-Cross Entropy;
- For Multiclass Classification, the common output activation is Softmax, and the common loss is Categorial Cross Entropy.


## 7.2 Gradient Descent & Backpropagation

### 7.2.1 Introduction

Gradient Descent updates parameters using gradients. The basic rule is as follows:

$$
θ_{new} = θ_{old} - η\frac{\partial{L}}{\partial{θ}}
$$

in which:

- $θ$ is any trainable parameter, such as weight and bias;
- $L$ is the Loss Function;
- $η$ is the learning rate (we caan view it as $Δx$, or the step length).

### 7.2.2 Chain Rule

We all know the chain rule. Here we use a light example to show it:

$$
z = wx, L = z^2
$$

if we want to update w:

$$
\frac{\partial L}{\partial w} = \frac{\partial L}{\partial z}\frac{\partial z}{\partial w}
$$

This is how backpropagation works: from the Loss to backpropagation the derivatives.

### 7.2.3 Code Implementation

Let's manually upadate the parameter:


```python
import numpy as np

# Input and target
x = np.array([1.0, 2.0])
y = 1.0

# Parameters
W1 = np.array(
    [
        [0.2, 0.4],
        [0.6, 0.1],
        [0.3, 0.5],
    ]
)
b1 = np.array([0.1, -0.1, 0.2])

W2 = np.array([[0.5, 0.2, 0.4]])
b2 = np.array([0.1])

learning_rate = 0.1

# Forward propagation
z1 = W1 @ x + b1
a1 = np.maximum(0, z1)

y_pred = W2 @ a1 + b2
loss_before = 0.5 * np.sum((y_pred - y) ** 2)

# Backpropagation
delta2 = y_pred - y

dW2 = np.outer(delta2, a1)
db2 = delta2.copy()

delta1 = (W2.T @ delta2) * (z1 > 0)

dW1 = np.outer(delta1, x)
db1 = delta1.copy()

# Gradient descent
W1 -= learning_rate * dW1
b1 -= learning_rate * db1
W2 -= learning_rate * dW2
b2 -= learning_rate * db2

# Forward propagation after update
z1_new = W1 @ x + b1
a1_new = np.maximum(0, z1_new)
y_pred_new = W2 @ a1_new + b2

loss_after = 0.5 * np.sum((y_pred_new - y) ** 2)

print("Loss before:", loss_before)
print("Loss after:", loss_after)

# Outer function turns the vectors into matrix, outer(a,b) produces a matrix whose shape is (a,b):
# import numpy as np
# a = np.array([1, 2, 3])
# b = np.array([4, 5, 6, 7])

# result = np.outer(a, b)
# print(result)

# the output is:
# [[ 4  5  6  7]
#  [ 8 10 12 14]
#  [12 15 18 21]]
```

Notice that after updating the Loss might not decrease, credit to the unsuitable learning rate.


### 7.2.4 Batch Gradient Descent & SGD & Mini-Batch

In reality, there are often massive samples. Thus we introduce different methods:

- Batch Gradient Descent: uses the entire training dataset as samples per update;
- Stochastic Gradient Descent: uses one sample as samples per update;
- Mini-Batch Gradient Descent: uses a small batch of samples as samples per update.

Modern Deep-learning training often uses mini-batch optimization.


## 7.3 Keras Sequential

### 7.3.1 Keras Training Workflow

The classic Keras training workflow is:

$$
Build \rightarrow Compile \rightarrow Fit \rightarrow Evaluate \rightarrow Predict
$$

### 7.3.2 Build Sequential and Dense Layers

Let's take a look at an example code:


```python
import os
import ssl

import tensorflow as tf

os.environ["SSL_CERT_FILE"] = "/etc/ssl/cert.pem"
ssl._create_default_https_context = lambda: ssl.create_default_context(
    cafile="/etc/ssl/cert.pem"
)

# 1. Load dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# 2. Flatten and normalize
X_train_norm = X_train.reshape(-1, 28 * 28) / 255.0
X_test_norm = X_test.reshape(-1, 28 * 28) / 255.0

model = tf.keras.models.Sequential()

model.add(
    tf.keras.layers.Dense(
        40,
        input_dim=28 * 28,
        activation="relu",
        name="Hidden1",
    )
)

model.add(
    tf.keras.layers.Dense(
        40,
        activation="relu",
        name="Hidden2",
    )
)

model.add(
    tf.keras.layers.Dense(
        10,
        activation="softmax",
        name="Output",
    )
)
```

The `units=40` in `Dense` implies the layer contains 40 neurons, not 40 samples nor imput features.

The trainable parameters of a dense layer is

$$
N = (n_{\text{inputs}}+1) \times n_{\text{units}}
$$

the extra $1$ stands for the bias of each neuron.


### 7.3.3 Compile

After building up sequential dense layers, we compile them:


```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
```

in which:

- `optimizer` defines how to update parameters;
- `loss` is the objective the training meant to improve;
- `metrics` is the metric to supervise or report performance.

The appropriate loss is related to the ground-truth format:

- Integer labels `[0,3,7]`: `sparse_categorical_crossentropy`;
- One-hot labels `[[1,0,0], [0,0,1]]`: `categorical_crossentropy`;
- Binary target `0/1`: `binary_crossentropy`.


### 7.3.4 Fit & Model Training

At this point, we can fit the training data with ease:


```python
model.fit(
    X_train_norm,
    y_train,
    epochs=4,  # completely roll through the training data four times.
    batch_size=32,  # use 32 samples in each update
    validation_split=0.25,  # split 25% of the input data for validation
    verbose=2,  # each epoch print out a report
)
```

We can bound this to a variable named history, which record the history for visualization:


```python
history = model.fit(
    X_train_norm,
    y_train,
    epochs=4,
    validation_split=0.25,
    verbose=2,
)

print(
    history.history.keys(),
    history.history["loss"],
    history.history["accuracy"],
    history.history["val_loss"],
    history.history["val_accuracy"],
)
```

### 7.3.5 Evaluate & Predict

As we have set `metrics=["accuracy"]` in the compilation, we can write:


```python
test_loss, test_accuracy = model.evaluate(
    X_test_norm,
    y_test,
    verbose=0,
)

print(test_loss, test_accuracy)
```

We can also find out the predicted class by using index calling:


```python
import numpy as np

probabilities = model.predict(X_test_norm)
print(probabilities.shape)

predicted_classes = np.argmax(
    probabilities,
    axis=1,  # find out the class with the maximum probability for each sample.
)
```

### 7.3.6 Complete Example: Fashion MNIST


```python
import numpy as np
import tensorflow as tf

# 1. Load dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# 2. Flatten and normalize
X_train = X_train.reshape(-1, 28 * 28) / 255.0
X_test = X_test.reshape(-1, 28 * 28) / 255.0

# 3. Build model
model = tf.keras.Sequential(
    [
        tf.keras.layers.Dense(
            40,
            activation="relu",
            input_shape=(784,),
        ),
        tf.keras.layers.Dense(
            40,
            activation="relu",
        ),
        tf.keras.layers.Dense(
            10,
            activation="softmax",
        ),
    ]
)

# 4. Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# 5. Train
history = model.fit(
    X_train,
    y_train,
    epochs=4,
    batch_size=32,
    validation_split=0.25,
    verbose=2,
)

# 6. Evaluate
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0,
)

print("Test Accuracy:", test_accuracy)

# 7. Predict
probabilities = model.predict(X_test[:5])
predictions = np.argmax(probabilities, axis=1)

print("Predictions:", predictions)
print("Ground Truth:", y_test[:5])
```
