## 7.1 Neural Networks

### 7.1.1 Neural Network Architecture

A classic neural network includes:

- Input Layer: receive Features;
- Hidden Layers: transform data with Weights, Biases and Activation Functions;
- Output Layer: produce the final output.

### 7.1.2 Forward Propagation

The calculation process of a neuron can be devided into 2 steps:

1. Linear Transformation: $z = w^T x +b$;
2. Activation Function: $a = f(z)$, in which f is the activation function.

Forward Propagation uses the output of a layer as the input of the next layer, repeat the process until the prediction comes out.

Let's take a look at an example:


```python
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


wi = np.array(
    [
        [0.25, 0.20],
        [0.30, 0.15],
    ]
)

wo = np.array(
    [
        [0.40, 0.45],
        [0.50, 0.55],
    ]
)

b1 = 0.30
b2 = 0.65
x = np.array([0.5, 0.1])

z1 = wi @ x + b1
a1 = sigmoid(z1)

z2 = wo @ a1 + b2
a2 = sigmoid(z2)

print("Hidden pre-activation:", z1)
print("Hidden activation:", a1)
print("Output pre-activation:", z2)
print("Final output:", a2)
```

notice: we agree on that:

$$
W.shape = (n_{\text{output neurons}},n_{\text{input neurons}})
$$


### 7.1.3 Activation Functions

Neural Networks need activation functions, because by linear algebra, the combination of several pure Linear layers is equal to a single Linear Layer. Thus, we need non-linear activation functions to learn more complex nonlinear relationships.

There are four important activation functions: Sigmoid, Tanh, ReLU and Softmax.


### 7.1.4 Loss Functions

Loss functions show the difference between model predictions and ground-truth tables.

- For Regression, the common output activation is Linear, and the common loss is MSE/MAE;
- For Binary Classification, the common output activation is Sigmoid, and the common loss is Binary-Cross Entropy;
- For Multiclass Classification, the common output activation is Softmax, and the common loss is Categorial Cross Entropy.


## 7.2 Gradient Descent & Backpropagation

### 7.2.1 Introduction

Gradient Descent updates parameters using gradients. The basic rule is as follows:

$$
θ_{new} = θ_{old} - η\frac{\partial{L}}{\partial{θ}}
$$

in which:

- $θ$ is any trainable parameter, such as weight and bias;
- $L$ is the Loss Function;
- $η$ is the learning rate (we caan view it as $Δx$, or the step length).

### 7.2.2 Chain Rule

We all know the chain rule. Here we use a light example to show it:

$$
z = wx, L = z^2
$$

if we want to update w:

$$
\frac{\partial L}{\partial w} = \frac{\partial L}{\partial z}\frac{\partial z}{\partial w}
$$

This is how backpropagation works: from the Loss to backpropagation the derivatives.

### 7.2.3 Code Implementation

Let's manually upadate the parameter:


```python
import numpy as np

# Input and target
x = np.array([1.0, 2.0])
y = 1.0

# Parameters
W1 = np.array(
    [
        [0.2, 0.4],
        [0.6, 0.1],
        [0.3, 0.5],
    ]
)
b1 = np.array([0.1, -0.1, 0.2])

W2 = np.array([[0.5, 0.2, 0.4]])
b2 = np.array([0.1])

learning_rate = 0.1

# Forward propagation
z1 = W1 @ x + b1
a1 = np.maximum(0, z1)

y_pred = W2 @ a1 + b2
loss_before = 0.5 * np.sum((y_pred - y) ** 2)

# Backpropagation
delta2 = y_pred - y

dW2 = np.outer(delta2, a1)
db2 = delta2.copy()

delta1 = (W2.T @ delta2) * (z1 > 0)

dW1 = np.outer(delta1, x)
db1 = delta1.copy()

# Gradient descent
W1 -= learning_rate * dW1
b1 -= learning_rate * db1
W2 -= learning_rate * dW2
b2 -= learning_rate * db2

# Forward propagation after update
z1_new = W1 @ x + b1
a1_new = np.maximum(0, z1_new)
y_pred_new = W2 @ a1_new + b2

loss_after = 0.5 * np.sum((y_pred_new - y) ** 2)

print("Loss before:", loss_before)
print("Loss after:", loss_after)

# Outer function turns the vectors into matrix, outer(a,b) produces a matrix whose shape is (a,b):
# import numpy as np
# a = np.array([1, 2, 3])
# b = np.array([4, 5, 6, 7])

# result = np.outer(a, b)
# print(result)

# the output is:
# [[ 4  5  6  7]
#  [ 8 10 12 14]
#  [12 15 18 21]]
```

Notice that after updating the Loss might not decrease, credit to the unsuitable learning rate.


### 7.2.4 Batch Gradient Descent & SGD & Mini-Batch

In reality, there are often massive samples. Thus we introduce different methods:

- Batch Gradient Descent: uses the entire training dataset as samples per update;
- Stochastic Gradient Descent: uses one sample as samples per update;
- Mini-Batch Gradient Descent: uses a small batch of samples as samples per update.

Modern Deep-learning training often uses mini-batch optimization.
