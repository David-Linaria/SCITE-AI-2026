### 5.1.2 Linear Regression Model

The simplest Linear Regression is called Simple Linear Regression, which has only one input feature:

$$
\hat{y} = wx + b
$$

in which $w$ is called weight/coefficient, and $b$ is called intercept/bias.


### 5.1.3 Residuals & Least Squares

Here are some definitions that are explained in the math class, we only rewrite to help review:

1. Residual is the deduction of actual value and predicted value, and Squared Residual is the square of Residual:

$$
e_i = y_i - \hat{y}_i \\
e_i^2 = (y_i - \hat{y}_i) ^ 2
$$

2. Ordinary Least Squares (OLS) is to find out the set of parameters which produces the smallest RSS:

$$
RSS = \sum_{i=1}^{n} (y_i - \hat{y}_i) ^ 2
$$

3. Mean Squared Error (MSE) can also indicate the sum of squares of residual:

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i) ^ 2
$$

Let's take a look at an example:


```python
import numpy as np

y_true = np.array([3, 5, 7])
y_pred = np.array([2, 6, 7])

residuals = y_true - y_pred
rss = np.sum(residuals**2)
mse = np.mean(residuals**2)

print("RSS:", rss)
print("MSE:", mse)
```

### 5.1.4 Implementing Linear Regression with Scikit-Learn

Here's a basic example:


```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Step 1: Prepare the dataset
X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
y = np.array([42, 51, 57, 66, 73, 80, 89, 96])

print("X shape:", X.shape)
print("y shape:", y.shape)
```

=> Notice that we have to reshape `X` to a two-dimensional array because sklearn wishes the Feature Matrix X to have $(n_{samples},\ n_{features})$.


```python
# Step 2: Train the model
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
)

model = LinearRegression()
model.fit(X_train, y_train)

# Step 3: Inspect the learned parameters
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)

# Step 4: Predict and evaluate
y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Actual Values:", y_test)

print("MSE:", mean_squared_error(y_test, y_pred))
print("R²:", r2_score(y_test, y_pred))
```

### 5.1.5 Regression Evaluation Metrics

For discrete data (classification), we use Acc, Prec, Recall, F1 as metrics. For Regression, we need different metrics:

$$
MAE = \frac{1}{n} |y_i - \hat{y}_i|,\ \ \ \
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i) ^ 2,\ \ \ \
RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i) ^ 2},\ \ \ \
R^2 = 1 - \frac{RSS}{TSS}
$$

Compared to MAE, MSE is more sensitive to greater Errors; Also notice that R2 can be negative.


```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

### 5.1.6 Visualizing the Regression Line

After linear regression, we can use Matplotlib to visualize them:


```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.scatter(X_train[:, 0], y_train, label="Training Data")
ax.scatter(X_test[:, 0], y_test, label="Test Data")

x_line = np.linspace(
    X.min(),
    X.max(),
    100,
).reshape(-1, 1)  # create continuous plotting cordinates
y_line = model.predict(x_line)

ax.plot(x_line[:, 0], y_line, color="red", label="Regression Line")

ax.set_title("Linear Regression")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Exam Score")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```
