### Mistake #001: The Shape of Array

In numpy, the dimension of an array is equivalent to the axis that can be indeendently indexded.


```python
import numpy as np

a = np.array([1, 2, 3])
print(a.shape)
print(a.ndim)

b = np.array([[1, 2, 3]])
print(b.shape)
print(b.ndim)

c = np.array([[1], [2], [3]])
print(c.shape)
print(c.ndim)
```

Why is the dimension of a 1 instead of 2? It is because we only need to index once to locate an element inside it. For example, `a[0]` refers to `1`.
